```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df = pd.read_csv(r"C:\Users\ASUS\OneDrive\Desktop\2025-Y2-S1_MLB-WE3G1-03 AI&ML\data\raw\diabetes_prediction_dataset.csv")

# Check for missing values
missing_info = df.isnull().sum()

if missing_info.sum() > 0:
    cols_with_missing = missing_info[missing_info > 0].index.tolist()
    df = df.drop(columns=cols_with_missing)
    print("Dropped columns with missing values:", cols_with_missing)
else:
    print("No missing values found.")

# Save the cleaned dataset
output_path = r"C:\Users\ASUS\OneDrive\Desktop\2025-Y2-S1_MLB-WE3G1-03 AI&ML\results\outputs\cleaned_diabetes_dataset.csv"
df.to_csv(output_path, index=False)
print(f"Cleaned dataset saved to: {output_path}")

# EDA Visualization - Correlation Heatmap
plt.figure(figsize=(10, 6))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap of Cleaned Dataset")
plt.show()

```

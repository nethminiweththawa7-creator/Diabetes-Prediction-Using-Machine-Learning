```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the cleaned dataset (after missing data handling)
df = pd.read_csv(r"C:\Users\ASUS\OneDrive\Desktop\2025-Y2-S1_MLB-WE3G1-03 AI&ML\results\outputs\cleaned_diabetes_dataset.csv")

# Step 1: Check categorical columns
print("Categorical columns:", df.select_dtypes(include=['object']).columns.tolist())

# Step 2: Encode categorical variables using One-Hot Encoding
df_encoded = pd.get_dummies(df, columns=['gender', 'smoking_history'], drop_first=True)

print("Dataset after encoding:\n", df_encoded.head())

# Step 3: Save encoded dataset
output_path = r"C:\Users\ASUS\OneDrive\Desktop\2025-Y2-S1_MLB-WE3G1-03 AI&ML\results\outputs\encoded_diabetes_dataset.csv"
df_encoded.to_csv(output_path, index=False)
print(f"Encoded dataset saved to: {output_path}")

# Step 4: EDA Visualization - Pie Chart of Gender (only Male & Female)
plt.figure(figsize=(6,6))

# Filter only Male and Female
gender_counts = df[df['gender'].isin(['Male', 'Female'])]['gender'].value_counts()

# Plot pie chart
plt.pie(gender_counts, labels=gender_counts.index, autopct='%1.1f%%',
        startangle=90, colors=['lightblue', 'lightpink'])
plt.title("Gender Distribution (Male vs Female)")
plt.show()


# Step 5: EDA Visualization - Countplot of Smoking History
plt.figure(figsize=(8,4))
sns.countplot(x='smoking_history', data=df, order=df['smoking_history'].value_counts().index)
plt.title("Smoking History Distribution")
plt.xticks(rotation=45)
plt.show()
```
